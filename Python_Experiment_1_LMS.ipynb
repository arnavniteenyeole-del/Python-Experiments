**6.Write a program to compute the length of the hypotenuse (c) of a right triangle using Pythagoras
theorem.**

```python
import math

a = float(input("Enter the length of side a: "))
b = float(input("Enter the length of side b: "))

c = math.sqrt(a**2 + b**2) #formula of hypotenuse is:root(a2+b2)

print("The length of the hypotenuse (c) is:", c)
```

**8. Write a program to find area of triangle when length of sides are given.**

```python
import math

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

s = (a + b + c) / 2  #SemiPerimeter
area = math.sqrt(s * (s - a) * (s - b) * (s - c))  #Herons's Formula

print("Area of the triangle is:", area)
```

**9.Write a program to convert given seconds into hours, minutes and remaining seconds.**

```python
total_seconds = int(input("Enter total seconds: "))

hours = total_seconds // 3600  #hours
remaining_seconds = total_seconds % 3600 #Remainder:Remaining Seconds

minutes = remaining_seconds // 60 #minutes
seconds = remaining_seconds % 60 #Remainder:Remaining Seconds

print("Hours:", hours)
print("Minutes:", minutes)
print("Seconds:", seconds)
```

**12.Write a program to print truth table for bitwise operators (&, | and ^ operators).**

```python
print("A B | A&B A|B A^B")
print("-------------------")

for A in [0, 1]:
    for B in [0, 1]:
        print(A, B, "|", A & B, " ", A | B, " ", A ^ B) #Use of Bitwise and, or and xor operators
```

**13.Write a program to find left shift and right shift values of a given number.**

```python
num = int(input("Enter a number: "))
shift = int(input("Enter number of positions to shift: "))

left_shift = num << shift  #Multiply's Number by (2 to the power of shift)
right_shift = num >> shift #Divide's the Number by (2 to the power of shift)

print("Original number:", num)
print("Left shift (num << shift):", left_shift)
print("Right shift (num >> shift):", right_shift)
```

**14.Using membership operator find whether a given number is in sequence (10,20,56,78,89).**

```python
T=(10,20,56,78,89)
num=int(input("Enter a Number:"))
if num in T:
  print(f"{num} is in sequence") #Printing number if found
else:
  print(f"{num} is not in sequence")
```
