**1.Check whether the given number is divisible by 3 and 5 both.**

```python
n = int(input("Enter a number: "))

if n % 3 == 0 and n % 5 == 0:
    print("Divisible by both 3 and 5")
else:
    print("Not divisible by both 3 and 5")
```

**3.Find the greatest among the two numbers. If numbers are equal than print “numbers are equal”.**

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("Greatest number is:", a)
elif b > a:
    print("Greatest number is:", b)
else:
    print("numbers are equal")
```

**5.Check whether the quadratic equation has real roots or imaginary roots. Display the roots.**

```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = b*b - 4*a*c   # discriminant

if d > 0:
    r1 = (-b + math.sqrt(d)) / (2*a)
    r2 = (-b - math.sqrt(d)) / (2*a)
    print("Roots are real and distinct")
    print("Root 1:", r1)
    print("Root 2:", r2)

elif d == 0:
    r = -b / (2*a)
    print("Roots are real and equal")
    print("Root:", r)

else:
    real = -b / (2*a)
    imag = math.sqrt(-d) / (2*a)
    print("Roots are imaginary (complex)")
    print("Root 1:", real, "+", imag, "i")
    print("Root 2:", real, "-", imag, "i")
```

**6.Find whether a given year is a leap year or not.**

```python
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

**7.Write a program which takes any date as input and display next date of the calendar
e.g.
I/P: day=20 month=9 year=2005
O/P: day=21 month=9 year 2005**

```python
day = int(input("Enter day: "))
month = int(input("Enter month: "))
year = int(input("Enter year: "))

# Check leap year
is_leap = (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)

# Days in each month
if month in [1, 3, 5, 7, 8, 10, 12]:
    max_days = 31
elif month in [4, 6, 9, 11]:
    max_days = 30
elif month == 2:
    max_days = 29 if is_leap else 28
else:
    print("Invalid month")
    exit()

# Find next date
if day < max_days:
    day += 1
else:
    day = 1
    if month == 12:
        month = 1
        year += 1
    else:
        month += 1

print("Next date:")
print("day =", day, "month =", month, "year =", year)
```

**8.Print the grade sheet of a student for the given range of cgpa. Scan marks of five subjects and calculate the percentage.
CGPA=percentage/10
CGPA range:
0 to 3.4 -> F
3.5 to 5.0->C+
5.1 to 6->B
6.1 to 7-> B+
7.1 to 8-> A
8.1 to 9->A+
9.1 to 10-> O (Outstanding)
Sample Gradesheet
Name: Rohit Sharma
Roll Number: R17234512 SAPID: 50005673
Sem: 1 Course: B.Tech. CSE AI&ML
Subject name: Marks
PDS: 70
Python: 80
Chemistry: 90
English: 60
Physics: 50
Percentage: 70%
CGPA:7.0
Grade:**

```python
name = input("Enter Name: ")
sapid = input("Enter SAPID: ")
sem = input("Enter Semester: ")
course = input("Enter Course: ")

print("\nEnter marks out of 100:")
DSA = int(input("DSA: "))
Python = int(input("Python: "))
Elements_of_AIML = int(input("Elements of AIML: "))
Digital_Electronics = int(input("Digital Electronics: "))
Physics = int(input("Physics: "))
AEM_II = int(input("AEM II: "))
EVS_II = int(input("EVS_II: "))
Time_Management = int(input("Time Management: "))

total = DSA+Python+Elements_of_AIML+Digital_Electronics+Physics+AEM_II+EVS_II+Time_Management
percentage = total / 8
cgpa = percentage / 10

# Grade calculation
if cgpa >= 0 and cgpa <= 3.4:
    grade = "F"
elif cgpa >= 3.5 and cgpa <= 5.0:
    grade = "C+"
elif cgpa >= 5.1 and cgpa <= 6.0:
    grade = "B"
elif cgpa >= 6.1 and cgpa <= 7.0:
    grade = "B+"
elif cgpa >= 7.1 and cgpa <= 8.0:
    grade = "A"
elif cgpa >= 8.1 and cgpa <= 9.0:
    grade = "A+"
elif cgpa >= 9.1 and cgpa <= 10:
    grade = "O (Outstanding)"
else:
    grade = "Invalid CGPA"

print("\n------ Grade Sheet ------")
print("Name:", name)
print(" SAPID:", sapid)
print("Sem:", sem, " Course:", course)
print("\nSubject name : Marks")
print("DSA:", DSA)
print("Python:", Python)
print("Elements of AIML:", Elements_of_AIML)
print("Digital Electronics:", Digital_Electronics)
print("Physics:", Physics)
print("AEM II:", AEM_II)
print("EVS II:", EVS_II)
print("Time Management:", Time_Management)

print("\nPercentage:", percentage, "%")
print("CGPA:", round(cgpa, 1))
print("Grade:", grade)
```
