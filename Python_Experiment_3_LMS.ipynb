***1.Find a factorial of given number.***

```python

n = int(input("Enter a number: "))

# Factorial is not defined for negative numbers
if n < 0:
    print("Factorial is not defined for negative numbers.")

else:
    # Initialize result to 1
    factorial = 1

    # Loop from 1 to n and multiply each number
    for i in range(1, n + 1):
        factorial = factorial * i

    print("Factorial of", n, "is:", factorial)
```

***2.Find whether the given number is Armstrong number.***

```python
num = int(input("Enter a number: "))

original_num = num

num_digits = len(str(num))

sum_of_powers = 0

# Loop through each digit
while num > 0:
    digit = num % 10            # Get last digit
    sum_of_powers += digit ** num_digits  # Raise digit to power of total digits
    num = num // 10            # Remove last digit

if sum_of_powers == original_num:
    print(original_num, "is an Armstrong number.")
else:
    print(original_num, "is not an Armstrong number.")
```

***4.Write a program to find if given number is prime number or not.***

```python
num = int(input("Enter a number: "))

# Prime numbers are greater than 1
if num <= 1:
    print(num, "is not a prime number.")

else:
    # Assume the number is prime initially
    is_prime = True

    # Check for factors from 2 to num-1
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break  # Exit loop if a factor is found

    # Print result
    if is_prime:
        print(num, "is a prime number.")
    else:
        print(num, "is not a prime number.")
```

***5.Check whether given number is palindrome or not.***

```python
num = int(input("Enter a number: "))

original_num = num

# Initialize reverse of the number
reverse_num = 0

# Reverse the number using a while loop
while num > 0:
    digit = num % 10              # Get the last digit
    reverse_num = reverse_num * 10 + digit  # Build the reversed number
    num = num // 10               # Remove the last digit

if original_num == reverse_num:
    print(original_num, "is a palindrome number.")
else:
    print(original_num, "is not a palindrome number.")
```

***6.Write a program to print sum of digits.***

```python
num = int(input("Enter a number: "))

# Make a copy of the number to work with
temp = num

# Initialize sum of digits
sum_of_digits = 0

# Loop to extract and add each digit
while temp > 0:
    digit = temp % 10          # Get the last digit
    sum_of_digits += digit    # Add digit to sum
    temp = temp // 10         # Remove the last digit

# Print the result
print("Sum of digits of", num, "is:", sum_of_digits)
```

***7.Count and print all numbers divisible by 5 or 7 between 1 to 100.***

```python
count = 0

print("Numbers divisible by 5 or 7 between 1 and 100:")

for num in range(1, 101):
    # Check if number is divisible by 5 or 7
    if num % 5 == 0 or num % 7 == 0:
        print(num, end=" ")
        count += 1

print("\n\nTotal count:", count)
```

***10.Write a program to print the following pattern
123454321
1234 *4321
123 * * 321
12 * * * 21
1 * * * * 1 .***

```python
n = 5  # Number of rows

for i in range(n):
    # Print increasing numbers from 1 to (n - i)
    for j in range(1, n - i + 1):
        print(j, end="")

    # Print stars in the middle
    if i > 0:
        print(" *" * i, end=" ")

    # Print decreasing numbers from (n - i) to 1
    for j in range(n - i, 0, -1):
        print(j, end="")

    # Move to next line
    print()
```

***11.Write a program to print the sum of the following series
1+ ½ + 1/3 + ¼ +….+1/n.***

```python
n = int(input("Enter the value of n: "))

# Initialize sum
series_sum = 0.0

# Check for valid input
if n <= 0:
    print("Please enter a positive integer.")
else:
    # Loop from 1 to n and add each term to the sum
    for i in range(1, n + 1):
        series_sum += 1 / i

    print("Sum of the series up to 1/", n, "is:", series_sum)
```
