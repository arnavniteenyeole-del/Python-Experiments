***1.Write a program to count and display the number of capital letters in a given string.***

```python
text = input("Enter a string: ")

# Initialize count of capital letters
capital_count = 0

# Loop through each character in the string
for ch in text:
    if ch.isupper():   # Check if character is uppercase
        capital_count += 1

print("Number of capital letters:", capital_count)
```

***2.Count total number of vowels in a given string.***

```python
text = input("Enter a string: ")

# Define vowels
vowels = "aeiouAEIOU"

vowel_count = 0

# Loop through each character in the string
for ch in text:
    if ch in vowels:
        vowel_count += 1

print("Total number of vowels:", vowel_count)
```

***4.WAP to enter a string and a substring. You have to print the number of times that the substring occurs in the given string. String traversal will take place from left to right, not from right to left.
Sample Input
ABCDCDC
CDC
Sample Output
2***

```python
string = input("Enter the main string: ")
substring = input("Enter the substring: ")

count = 0
start = 0

while start <= len(string) - len(substring):
    # Check if substring matches at the current position
    if string[start:start + len(substring)] == substring:
        count += 1
    start += 1  # Move one step to the right

print(count)
```

***5.Given a string containing both upper and lower case alphabets. Write a Python program to count the number of occurrences of each alphabet (case insensitive) and display the same.
Sample Input
ABaBCbGc
Sample Output
2A
3B
2C
1G***

```python
text = input("Enter a string: ")

# Convert the string to uppercase
text = text.upper()

count_dict = {}

# Count occurrences of each alphabet
for ch in text:
    if ch.isalpha():  # Check if character is an alphabet
        if ch in count_dict:
            count_dict[ch] += 1
        else:
            count_dict[ch] = 1

# Print the result sorted alphabetically
for key in sorted(count_dict.keys()):
    print(count_dict[key], key, sep="")
```

***6.Program to count number of unique words in a given sentence using sets.***

```python
sentence = input("Enter a sentence: ")

# Convert sentence to lowercase
sentence = sentence.lower()

# Split the sentence into words
words = sentence.split()

# Convert list of words to a set to remove duplicates
unique_words = set(words)

# Print the unique words and their count
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```

***7.Create 2 sets s1 and s2 of n fruits each by taking input from user and find:
a)
Fruits which are in both sets s1 and s2
b)
Fruits only in s1 but not in s2
c)
Count of all fruits from s1 and s2***

```python
n = int(input("Enter number of fruits for each set: "))

s1 = set()
s2 = set()

# Input fruits for set s1
print("Enter fruits for set s1:")
for i in range(n):
    fruit = input(f"Fruit {i+1}: ")
    s1.add(fruit)

# Input fruits for set s2
print("\nEnter fruits for set s2:")
for i in range(n):
    fruit = input(f"Fruit {i+1}: ")
    s2.add(fruit)

print("\nSet s1:", s1)
print("Set s2:", s2)

# a) Fruits in both sets
common_fruits = s1.intersection(s2)
print("\n(a) Fruits in both s1 and s2:", common_fruits)

# b) Fruits only in s1 but not in s2
only_in_s1 = s1.difference(s2)
print("(b) Fruits only in s1 but not in s2:", only_in_s1)

# c) Count of all fruits from s1 and s2 (unique fruits)
all_fruits = s1.union(s2)
print("(c) Count of all unique fruits from s1 and s2:", len(all_fruits))
```

***8.Take two sets and apply various set operations on them :
S1 = {Red ,yellow, orange , blue }
S2 = {violet, blue , purple}***

```python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}

# Print the original sets
print("Set S1:", S1)
print("Set S2:", S2)

# Union: elements in either S1 or S2
union_set = S1.union(S2)
print("\nUnion of S1 and S2:", union_set)

# Intersection: common elements in both sets
intersection_set = S1.intersection(S2)
print("Intersection of S1 and S2:", intersection_set)

# Difference: elements in S1 but not in S2
difference_S1_S2 = S1.difference(S2)
print("Difference of S1 - S2:", difference_S1_S2)

# Difference: elements in S2 but not in S1
difference_S2_S1 = S2.difference(S1)
print("Difference of S2 - S1:", difference_S2_S1)

# Symmetric Difference: elements in either set but not in both
symmetric_difference = S1.symmetric_difference(S2)
print("Symmetric Difference of S1 and S2:", symmetric_difference)
```
