***1.Scan n values in range 0-3 and print the number of times each value has occurred.***

```python
n = int(input("Enter number of values: "))

# Initialize counters for 0, 1, 2, 3
count_0 = 0
count_1 = 0
count_2 = 0
count_3 = 0

print("Enter values (only 0, 1, 2, or 3):")

# Take n values from the user
for i in range(n):
    val = int(input(f"Value {i+1}: "))

    if val == 0:
        count_0 += 1
    elif val == 1:
        count_1 += 1
    elif val == 2:
        count_2 += 1
    elif val == 3:
        count_3 += 1
    else:
        print("Invalid input! Please enter only 0, 1, 2, or 3.")

# Print the counts
print("\nOccurrences:")
print("0 occurred", count_0, "times")
print("1 occurred", count_1, "times")
print("2 occurred", count_2, "times")
print("3 occurred", count_3, "times")
```

***2.Program to create a tuple of n numeric values and find their average.***

```python
n = int(input("Enter number of values: "))

# Create an empty list first (since tuples are immutable)
values_list = []

# Take n numeric inputs from the user
for i in range(n):
    num = float(input(f"Enter value {i+1}: "))
    values_list.append(num)

# Convert the list to a tuple
values_tuple = tuple(values_list)

# Calculate the average
if n > 0:
    average = sum(values_tuple) / n
    print("Tuple of values:", values_tuple)
    print("Average of values:", average)
else:
    print("No values entered. Average cannot be calculated.")
```

***3.WAP to input a list of scores for N students in a list data type. Find the score of the runner-up and print the output.
Sample Input
N = 5
Scores= 2 3 6 6 5
Sample output
5
Note: Given list is [2, 3, 6, 6, 5]. The maximum score is 6, second maximum is 5. Hence, we print 5 as the runner-up score.***

```python
N = int(input("Enter number of students: "))

# Take scores as space-separated input
scores = list(map(int, input("Enter the scores: ").split()))

# Convert list to set to remove duplicate scores
unique_scores = set(scores)

# Remove the maximum score
max_score = max(unique_scores)
unique_scores.remove(max_score)

# The next maximum is the runner-up score
runner_up = max(unique_scores)

# Print the runner-up score
print("Runner-up score:", runner_up)
```

***4.Create a dictionary of n persons where key is name and value is city.
a) Display all names
b) Display all city names
c) Display student name and city of all students.
d) Count number of students in each city.***

```python
n = int(input("Enter number of students: "))

# Create empty dictionary
students = {}

# Input name and city for each student
for i in range(n):
    name = input(f"Enter name of student {i+1}: ")
    city = input(f"Enter city of student {i+1}: ")
    students[name] = city

# a) Display all names
print("\n(a) All student names:")
for name in students.keys():
    print(name)

# b) Display all city names
print("\n(b) All city names:")
for city in students.values():
    print(city)

# c) Display student name and city of all students
print("\n(c) Student name and city:")
for name, city in students.items():
    print(name, "->", city)

# d) Count number of students in each city
print("\n(d) Number of students in each city:")
city_count = {}

for city in students.values():
    if city in city_count:
        city_count[city] += 1
    else:
        city_count[city] = 1

for city, count in city_count.items():
    print(city, ":", count)
```

***5.Store details of n movies in a dictionary by taking input from the user. Each movie must store details like name, year, director name, production cost, collection made (earning) & perform the following :-
a)
print all movie details
b)
display name of movies released before 2015
c)
print movies that made a profit.
d)
print movies directed by a particular director.***

```python
n = int(input("Enter number of movies: "))

# Dictionary to store movie details
# Key: movie name, Value: dictionary of details
movies = {}

# Input movie details
for i in range(n):
    print(f"\nEnter details of movie {i+1}:")
    name = input("Movie name: ")
    year = int(input("Release year: "))
    director = input("Director name: ")
    cost = float(input("Production cost: "))
    collection = float(input("Collection made (earning): "))

    movies[name] = {
        "year": year,
        "director": director,
        "cost": cost,
        "collection": collection
    }

# a) Print all movie details
print("\n(a) All movie details:")
for name, details in movies.items():
    print("\nMovie Name:", name)
    print("Year:", details["year"])
    print("Director:", details["director"])
    print("Production Cost:", details["cost"])
    print("Collection:", details["collection"])

# b) Display names of movies released before 2015
print("\n(b) Movies released before 2015:")
for name, details in movies.items():
    if details["year"] < 2015:
        print(name)

# c) Print movies that made a profit (collection > cost)
print("\n(c) Movies that made a profit:")
for name, details in movies.items():
    if details["collection"] > details["cost"]:
        print(name)

# d) Print movies directed by a particular director
search_director = input("\nEnter director name to search: ")
print(f"\n(d) Movies directed by {search_director}:")
found = False
for name, details in movies.items():
    if details["director"].lower() == search_director.lower():
        print(name)
        found = True

if not found:
    print("No movies found for this director.")
```

***6.Create a contact book where users can store, search, update, and delete contacts. Use dictionary for storing contacts.***

```python
# Dictionary to store contacts
# Key: Name, Value: Phone number
contacts = {}

def add_contact():
    name = input("Enter name: ")
    phone = input("Enter phone number: ")
    contacts[name] = phone
    print("Contact added successfully!")

def search_contact():
    name = input("Enter name to search: ")
    if name in contacts:
        print("Name:", name)
        print("Phone:", contacts[name])
    else:
        print("Contact not found.")

def update_contact():
    name = input("Enter name to update: ")
    if name in contacts:
        new_phone = input("Enter new phone number: ")
        contacts[name] = new_phone
        print("Contact updated successfully!")
    else:
        print("Contact not found.")

def delete_contact():
    name = input("Enter name to delete: ")
    if name in contacts:
        del contacts[name]
        print("Contact deleted successfully!")
    else:
        print("Contact not found.")

def display_contacts():
    if not contacts:
        print("Contact book is empty.")
    else:
        print("\nAll Contacts:")
        for name, phone in contacts.items():
            print(name, ":", phone)

# Menu-driven program
while True:
    print("\n--- Contact Book Menu ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Display All Contacts")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    if choice == '1':
        add_contact()
    elif choice == '2':
        search_contact()
    elif choice == '3':
        update_contact()
    elif choice == '4':
        delete_contact()
    elif choice == '5':
        display_contacts()
    elif choice == '6':
        print("Exiting Contact Book. Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```

***7.Create a Todo list Manager where users can add, view, and remove tasks. Use List for storing tasks.***

```python
# List to store tasks
tasks = []

def add_task():
    task = input("Enter a new task: ")
    tasks.append(task)
    print("Task added successfully!")

def view_tasks():
    if not tasks:
        print("Your todo list is empty.")
    else:
        print("\nYour Tasks:")
        for i, task in enumerate(tasks, start=1):
            print(f"{i}. {task}")

def remove_task():
    if not tasks:
        print("No tasks to remove.")
        return

    view_tasks()
    try:
        index = int(input("Enter the task number to remove: "))
        if 1 <= index <= len(tasks):
            removed = tasks.pop(index - 1)
            print("Removed task:", removed)
        else:
            print("Invalid task number.")
    except ValueError:
        print("Please enter a valid number.")

# Menu-driven program
while True:
    print("\n--- Todo List Manager ---")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Remove Task")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        add_task()
    elif choice == '2':
        view_tasks()
    elif choice == '3':
        remove_task()
    elif choice == '4':
        print("Exiting Todo List Manager. Bye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
