***1.Write a Python function to find the maximum and minimum numbers from a sequence
of numbers.(Note: Do not use built-in functions.)***

```python
def find_max_min(numbers):
    # Assume the first element is both max and min initially
    maximum = numbers[0]
    minimum = numbers[0]

    # Traverse the sequence starting from the second element
    for num in numbers[1:]:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num

    return maximum, minimum


# Example usage
nums = list(map(int, input("Enter numbers separated by space: ").split()))

max_val, min_val = find_max_min(nums)

print("Maximum:", max_val)
print("Minimum:", min_val)
```

***2.Write a Python function that takes a positive integer and returns the sum of the cube of
all the positive integers smaller than the specified number.***

```python
def sum_of_cubes(n):
    total = 0

    # Loop from 1 to n-1
    for i in range(1, n):
        total += i ** 3

    return total


# Example usage
num = int(input("Enter a positive integer: "))
result = sum_of_cubes(num)
print("Sum of cubes of numbers smaller than", num, "is:", result)
```

***3.Write a Python function to print 1 to n using recursion.(Note: Do not use loop)***

```python
def print_1_to_n(n):
    # Base case: if n is 0, stop the recursion
    if n == 0:
        return
    # First print numbers from 1 to n-1
    print_1_to_n(n - 1)
    # Then print n
    print(n)

# Example usage
num = int(input("Enter a positive integer: "))
print_1_to_n(num)
```

***4.Write a recursive function to print Fibonacci series upto n terms.***

```python
# Recursive function to return the nth Fibonacci number
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

# Function to print Fibonacci series up to n terms using recursion
def print_fibonacci(n, i=0):
    if i == n:          # Base case: printed n terms
        return
    print(fib(i), end=" ")
    print_fibonacci(n, i + 1)

# Example usage
num = int(input("Enter number of terms: "))
print("Fibonacci series:")
print_fibonacci(num)
```

***5.Write a lambda function to find volume of cone.***

```python
# Lambda function to calculate volume of a cone
volume_of_cone = lambda r, h: (1/3) * 3.14159 * r * r * h

# Example usage
radius = float(input("Enter radius of the cone: "))
height = float(input("Enter height of the cone: "))

volume = volume_of_cone(radius, height)
print("Volume of the cone:", volume)
```

***6.Write a lambda function which gives tuple of max and min from a list.
Sample input: [10, 6, 8, 90, 12, 56]
Sample output: (90,6)***

```python
# Lambda function to return (max, min) from a list
max_min = lambda lst: (max(lst), min(lst))

# Example usage
numbers = [10, 6, 8, 90, 12, 56]
result = max_min(numbers)

print(result)   # Output: (90, 6)
```

***7.Write functions to explain the mentioned concepts:
a. Keyword argument
b. Default argument
c. Variable-length argument***

**a)Keyword Argument**

```python
# Function using keyword arguments
def student_info(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

# Calling the function using keyword arguments
student_info(age=20, city="Pune", name="Rahul")
```

**b)Default Argument**

```python
# Function with default arguments
def greet(name, message="Hello"):
    print(message, name)

# Calling with both arguments
greet("Anjali", "Welcome")

# Calling with default argument
greet("Anjali")
```

**c)Variable-Length Argument**

```python
# Function with variable-length arguments
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    print("Sum:", total)

# Calling the function with different number of arguments
add_numbers(10, 20)
add_numbers(1, 2, 3, 4, 5)
```

***8.Write a program to check whether all the values in a dictionary are the same or not using
a lambda function.***

```python
# Input dictionary
data = {
    "a": 10,
    "b": 10,
    "c": 10,
    "d": 10
}

# Lambda function to check if all values are the same
all_same = lambda d: len(set(d.values())) == 1

# Check and print result
if all_same(data):
    print("All values in the dictionary are the same.")
else:
    print("Values in the dictionary are not the same.")
```

***9.Write a program to create two lists and generate a dictionary with keys from list1 and
values from list2.***

```python
n = int(input("Enter number of elements: "))

# Create first list (keys)
list1 = []
print("Enter elements for list1 (keys):")
for i in range(n):
    key = input(f"Key {i+1}: ")
    list1.append(key)

# Create second list (values)
list2 = []
print("\nEnter elements for list2 (values):")
for i in range(n):
    value = input(f"Value {i+1}: ")
    list2.append(value)

# Generate dictionary using zip()
result_dict = dict(zip(list1, list2))

# Print the dictionary
print("\nGenerated Dictionary:")
print(result_dict)
```
